```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df=pd.read_csv("cleaned_data.csv")
```

```python
sns.countplot(x='is_fraud', data=df)
plt.title("Fraud vs Non-Fraud Count")
plt.savefig("fraud vs non fraud count.jpg")
plt.show()

```

```python
sns.histplot(data=df, x='transaction_amount', hue='is_fraud', bins=50, kde=True)
plt.title("Transaction Amount Distribution by Fraud")
plt.savefig("Transaction Amount Distribution by Fraud.jpg")
plt.show()

```

```python
sns.histplot(data=df, x='transaction_time', hue='is_fraud', bins=24, kde=False)
plt.title("Transaction Time vs Fraud")
plt.xlabel("Hour of Day (0–23)")
plt.savefig("Transaction Time vs Fraud.jpg")
plt.show()

```

```python
device_type_map = {
    0: "POS",
    1: "Tablet",
    2: "Desktop",
    3: "Mobile"
}
plt.figure(figsize=(8, 5))
sns.countplot(x='device_type', hue='is_fraud', data=df)
plt.title("Device Type vs Fraud")

plt.xticks(ticks=range(len(device_type_map)), labels=[device_type_map[i] for i in range(len(device_type_map))])
plt.xlabel("Device Type")
plt.ylabel("Transaction Count")
plt.legend(title="Is Fraud", labels=["Not Fraud (0)", "Fraud (1)"])

plt.tight_layout()
plt.savefig("Device_Type_vs_Fraud.jpg")
plt.show()
```

```python
channel_map = {
    0: "POS",
    1: "Online"
}

plt.figure(figsize=(8, 5))
sns.countplot(x='channel', hue='is_fraud', data=df)

plt.title("Channel vs Fraud", fontsize=14)
plt.xlabel("Channel", fontsize=12)
plt.ylabel("Transaction Count", fontsize=12)

plt.xticks(ticks=range(len(channel_map)), labels=[channel_map[i] for i in range(len(channel_map))])

plt.legend(title="Is Fraud", labels=["Not Fraud (0)", "Fraud (1)"])

plt.tight_layout()
plt.savefig("Channel_vs_Fraud.jpg")
plt.show()
```

```python
merchant_map = {
    0: "Grocery",
    1: "Electronics",
    2: "Clothing",
    3: "Travel",
    4: "Food",
    5: "Entertainment"
}

plt.figure(figsize=(10, 6))
sns.countplot(x='merchant_category', hue='is_fraud', data=df)

plt.title("Merchant Category vs Fraud", fontsize=14)
plt.xlabel("Merchant Category", fontsize=12)
plt.ylabel("Transaction Count", fontsize=12)

plt.xticks(ticks=range(len(merchant_map)), labels=[merchant_map[i] for i in range(len(merchant_map))])

plt.legend(title="Is Fraud", labels=["Not Fraud (0)", "Fraud (1)"])

plt.tight_layout()
plt.savefig("Merchant_Category_vs_Fraud.jpg")
plt.show()
```
