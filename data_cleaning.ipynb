```python
df=df.drop(['transaction_id'],axis=1)
```

```python
df.head()
```

```python
df['device_type']=df['device_type'].str.lower()
df['card_present']=df['card_present'].str.lower()
df['channel']=df['channel'].str.lower()
df['merchant_category']=df['merchant_category'].str.lower()
```

```python
df.head()
```

```python
# Fix for card_present
df['card_present'] = df['card_present'].str.lower().map({"yes": 1, "no": 0})

# Fix for device_type
df['device_type'] = df['device_type'].str.lower().map({"pos": 0, "tablet": 1, "desktop": 2, "mobile": 3})
```

```python
df.head()
```

```python
df=df.drop(['location'],axis=1)
```

```python
df.head()
```

```python
df['merchant_category'] = df['merchant_category'].str.lower().map({
    "grocery": 0,
    "electronics": 1,
    "clothing": 2,
    "travel": 3,
    "food": 4,
    "entertainment": 5
})
```

```python
df['channel'] = df['channel'].str.lower().map({
    "pos": 0,
    "online": 1
})
```

```python
df.head()
```

```python
df.to_csv("cleaned_data.csv",index=False)
```
